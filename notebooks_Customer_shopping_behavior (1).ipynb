```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ','_')
df= df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age group
labels=['Young Adult', 'Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' :7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping)
```

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Read your dataset
df = pd.read_csv('customer_shopping_behavior.csv')

# 2. Connect to an IN-MEMORY SQLite database (bypasses the virtual disk entirely)
engine = create_engine("sqlite://")

# 3. Save your DataFrame 
df.to_sql("customer", engine, if_exists="replace", index=False)

print("🎉 Success! The data is safely loaded into an in-memory database.")
```

```python
# Run your SQL query directly inside the notebook using pandas
query = "SELECT * FROM customer LIMIT 20"
pd.read_sql(query, engine)
```

```python
# Save your cleaned dataframe to a physical CSV file in your workspace
df.to_csv('cleaned_customer_data.csv', index=False)
print("File 'cleaned_customer_data.csv' successfully created!")
```
